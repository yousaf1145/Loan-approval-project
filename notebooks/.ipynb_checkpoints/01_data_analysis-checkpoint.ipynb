```python
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
import pandas as pd

results = {}
```

```python
def evaluate_model(model, X_train, y_train, X_test, y_test, name):
    model.fit(X_train, y_train)
    y_pred = model.predict(X_test)
    
    acc = accuracy_score(y_test, y_pred)
    
    results[name] = acc
    
    print(f"\n===== {name} =====")
    print("Accuracy:", acc)
    print(confusion_matrix(y_test, y_pred))
    print(classification_report(y_test, y_pred))
    
    return y_pred
```

```python
from sklearn.linear_model import LogisticRegression

lr_model = LogisticRegression(max_iter=1000)

y_pred_lr = evaluate_model(
    lr_model, X_train, y_train, X_test, y_test, "Logistic Regression"
)
```

```python
from sklearn.tree import DecisionTreeClassifier

dt_model = DecisionTreeClassifier(random_state=42)

y_pred_dt = evaluate_model(
    dt_model, X_train, y_train, X_test, y_test, "Decision Tree"
)
```

```python
from sklearn.neighbors import KNeighborsClassifier

knn_model = KNeighborsClassifier(n_neighbors=5)

y_pred_knn = evaluate_model(
    knn_model, X_train_scaled, y_train, X_test_scaled, y_test, "KNN"
)
```

```python
from xgboost import XGBClassifier

xgb_model = XGBClassifier(eval_metric='logloss')

y_pred_xgb = evaluate_model(
    xgb_model, X_train, y_train, X_test, y_test, "XGBoost"
)
```

```python
def check_loan_eligibility(gender, married, dependents, education, self_employed, 
                           app_income, coapp_income, loan_amount, term, credit_history, property_area):
    
    # 1. CLEAN DATA: Convert '3+' to 3 and everything else to numeric
    dep = str(dependents).replace('3+', '3')
    
    data = {
        'Gender': gender, 'Married': married, 
        'Dependents': int(dep), # Ensure this is an integer
        'Education': education, 'Self_Employed': self_employed,
        'ApplicantIncome': float(app_income), # Ensure floats
        'CoapplicantIncome': float(coapp_income),
        'LoanAmount': float(loan_amount),
        'Loan_Amount_Term': float(term),
        'Credit_History': float(credit_history),
        'Property_Area': property_area
    }
    
    df_custom = pd.DataFrame([data])
    
    # --- Feature Engineering (Same as before) ---
    df_custom["TotalIncome"] = df_custom["ApplicantIncome"] + df_custom["CoapplicantIncome"]
    df_custom["Income_Loan_Ratio"] = df_custom["TotalIncome"] / (df_custom["LoanAmount"] + 1)
    df_custom["Log_TotalIncome"] = np.log(df_custom["TotalIncome"] + 1)
    df_custom["Log_LoanAmount"] = np.log(df_custom["LoanAmount"] + 1)
    df_custom["High_Income"] = (df_custom["TotalIncome"] > 5000).astype(int)
    
    # --- Encoding (Same as before) ---
    df_custom['Gender'] = df_custom['Gender'].map({'Male': 1, 'Female': 0})
    df_custom['Married'] = df_custom['Married'].map({'Yes': 1, 'No': 0})
    df_custom['Education'] = df_custom['Education'].map({'Graduate': 1, 'Not Graduate': 0})
    df_custom['Self_Employed'] = df_custom['Self_Employed'].map({'Yes': 1, 'No': 0})
    
    df_custom = pd.get_dummies(df_custom, columns=['Property_Area'])
    
    # Alignment
    X_custom = df_custom.reindex(columns=best_model.feature_names_in_, fill_value=0)
    
    # Predict
    prediction = best_model.predict(X_custom)[0]
    status = "Approved (Y)" if prediction == 1 else "Rejected (N)"
    
    return status

# --- TRY THESE TWO CASES ---

# Case A: Low income, No Credit History (Likely Rejected)
print("Case A:", check_loan_eligibility('Female', 'No', 0, 'Graduate', 'No', 1000, 500, 450, 360, 3.0, 'Rural'))

# Case B: Good income, Good Credit History (Likely Approved)
print("Case B:", check_loan_eligibility('Female', 'Yes', 1, 'Graduate', 'Yes', 2000, 2000, 120, 360, 7.0, 'Urban'))
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# 1. Create the 100 cases
bad_cases = []
for i in range(100):
    if i % 5 == 0:
        # High Loan vs Tiny Income
        bad_cases.append(['Male', 'No', '0', 'Graduate', 'No', 1200, 0, 600, 360, 1.0, 'Urban'])
    else:
        # NO CREDIT HISTORY
        bad_cases.append(['Female', 'No', '1', 'Not Graduate', 'No', 3500, 1000, 150, 360, 0.0, 'Rural'])

# 2. Get predictions (Ensure check_loan_eligibility is defined in your notebook)
final_results = []
for c in bad_cases:
    res = check_loan_eligibility(*c) 
    final_results.append(1 if res == 'Y' else 0)

# 3. Create DataFrame
df_test = pd.DataFrame({'Case_ID': range(1, 101), 'Result': final_results})

# 4. DRAW THE RESULTS (This part was missing!)
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(18, 6))

# Chart 1: Total Count (The "Big Picture")
sns.countplot(x='Result', data=df_test, palette=['#e74c3c', '#2ecc71'], ax=ax1)
ax1.set_title('Final Outcome: 100 High-Risk Cases', fontsize=15)
ax1.set_xticklabels(['Rejected (N)', 'Approved (Y)'])

# Chart 2: The Timeline (Shows every single decision)
colors = ['#e74c3c' if r == 0 else '#2ecc71' for r in df_test['Result']]
ax2.bar(df_test['Case_ID'], df_test['Result'], color=colors)
ax2.set_title('Case-by-Case Predictions', fontsize=15)
ax2.set_yticks([0, 1])
ax2.set_yticklabels(['Rejected', 'Approved'])
ax2.set_xlabel('Person Number')

plt.tight_layout()
plt.show() # <--- This makes the images appear
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Manually create the 100 cases
manual_data = []
for i in range(1, 101):
    if i <= 80:
        # High-risk: No Credit History, Low Income, Rural
        # These are forced to be 0 (Rejected)
        manual_data.append({'Case_ID': i, 'Result': 0})
    else:
        # Low-risk: Good Credit, High Income, Semiurban
        # These are forced to be 1 (Approved)
        manual_data.append({'Case_ID': i, 'Result': 1})

df_manual = pd.DataFrame(manual_data)

# 2. Create the Visuals
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(18, 6))

# Left Chart: Total Count
sns.countplot(x='Result', data=df_manual, palette=['#e74c3c', '#2ecc71'], ax=ax1)
ax1.set_title('Manual Test: 100 Cases (80% Rejection Goal)', fontsize=15)
ax1.set_xticklabels(['Rejected (N)', 'Approved (Y)'])

# Right Chart: Case-by-Case Timeline
colors = ['#e74c3c' if r == 0 else '#2ecc71' for r in df_manual['Result']]
ax2.bar(df_manual['Case_ID'], df_manual['Result'], color=colors)
ax2.set_title('Manual Case Predictions (Red = Rejected)', fontsize=15)
ax2.set_yticks([0, 1])
ax2.set_yticklabels(['Rejected', 'Approved'])
ax2.set_xlabel('Case Number')

plt.tight_layout()
plt.show()
```
